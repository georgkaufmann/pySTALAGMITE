# `pySTALAGMITE`

pySTALAGMITE, a program package prediction the shape of a stalagmite from pre-defined climate conditions.

----
# Flow chart for program `pySTALAGMITE`

<img src='images/pySTALAGMITE_Flowchart.png' style='width:12cm;float: right;'>

## Read

In the first section, we want to read in all relevant parameter values from files:
- `readParameter2D()`: Reads in the **geometry** of the modelling domain.
- `readTimeline2D()`: Reads in the **boundary conditions** for fixed head and flow conditions for the model.

## Create

In the second section, we want to create the grid setup, set boundary conditions, define properties:
- `createGrid2D()`: Creates the **coordinates** of the modelling domain.

## Grow
In the third section, the actual time-dependent calculations are performed:
- `setClimate()`: set climate conditions for time step.
- `CEQ_limestone_()`: calculate calcium equilibrium concentration (open, closed system).
- `growthrate2D()`: Calculate growth rate at apex.
- `equiRadius2D()`: Calculate equilibrium radius.
- `refineGrid2D()`: Refine grid, if necessary.

## Final
In this last sections, final tasks can be embedded, e.g. plotting of the results.
